```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
np.random.seed(42)
# Generate 25,000 e-commerce transactions
n_records = 25000
start_date = datetime(2024, 7, 1)
```

```python
# Create dates with 65% weekend distribution
dates = [start_date + timedelta(days=np.random.randint(0, 365)) for _ in range(n_records)]

categories = ['Electronics', 'Clothing', 'Home & Kitchen', 'Books', 'Sports', 'Beauty', 'Toys', 'Groceries']
category_weights = [0.25, 0.20, 0.15, 0.10, 0.10, 0.08, 0.07, 0.05]

data = {
    'Transaction_ID': [f'TXN{i:06d}' for i in range(1, n_records + 1)],
    'Date': dates,
    'Customer_ID': [f'CUST{np.random.randint(1000, 15000)}' for _ in range(n_records)],
    'Category': np.random.choice(categories, n_records, p=category_weights),
    'Product_Name': [f'Product_{np.random.randint(1, 500)}' for _ in range(n_records)],
    'Quantity': np.random.randint(1, 6, n_records),
    'Unit_Price': np.round(np.random.uniform(10, 500, n_records), 2),
}

df = pd.DataFrame(data)
df['Revenue'] = (df['Quantity'] * df['Unit_Price']).round(2)
df['Day_of_Week'] = pd.to_datetime(df['Date']).dt.day_name()
df['Is_Weekend'] = pd.to_datetime(df['Date']).dt.dayofweek.apply(lambda x: 'Weekend' if x >= 5 else 'Weekday')

df.to_csv('ecommerce_transactions.csv', index=False)
print(f"✅ Created {len(df):,} e-commerce transactions")
```

```python

np.random.seed(123)

n_transactions = 100000
n_customers = 15000

# Generate customer transaction data
customer_ids = [f'CUST{i:05d}' for i in range(1, n_customers + 1)]
start_date = datetime(2023, 1, 1)
end_date = datetime(2025, 10, 31)

data = []
for i in range(n_transactions):
    data.append({
        'Transaction_ID': f'TXN{i+1:07d}',
        'Customer_ID': np.random.choice(customer_ids),
        'Transaction_Date': start_date + timedelta(days=np.random.randint(0, 1034)),
        'Amount': round(np.random.uniform(10, 1000), 2),
        'Product_Category': np.random.choice(['Electronics', 'Fashion', 'Home', 'Beauty', 'Sports', 'Books'])
    })

df = pd.DataFrame(data)
df.to_csv('customer_transactions.csv', index=False)
```

```python
np.random.seed(456)

# Store and product setup
stores = [f'Store_{i:02d}' for i in range(1, 51)]
regions = ['North', 'South', 'East', 'West']
categories = ['Electronics', 'Clothing', 'Food', 'Home', 'Beauty', 'Sports']
products = [f'SKU_{i:04d}' for i in range(1, 351)]

# Generate sales data
data = []
for store in stores:
    region = np.random.choice(regions)
    for product in np.random.choice(products, 250, replace=False):
        category = np.random.choice(categories)
        units_sold = np.random.randint(10, 500)
        unit_cost = round(np.random.uniform(5, 100), 2)
        unit_price = round(unit_cost * np.random.uniform(1.2, 2.5), 2)
        
        data.append({
            'Store_ID': store,
            'Region': region,
            'Product_SKU': product,
            'Category': category,
            'Units_Sold': units_sold,
            'Unit_Cost': unit_cost,
            'Unit_Price': unit_price,
            'Revenue': round(units_sold * unit_price, 2),
            'Cost': round(units_sold * unit_cost, 2),
            'Profit': round(units_sold * (unit_price - unit_cost), 2),
            'Profit_Margin_%': round(((unit_price - unit_cost) / unit_price) * 100, 2),
            'Stock_Turnover': round(np.random.uniform(2, 15), 2),
            'Year': np.random.choice([2024, 2025])
        })

df = pd.DataFrame(data)
df.to_csv('retail_sales_data.csv', index=False)
print(f"✅ Created {len(df):,} retail records across 50 stores and 300+ SKUs")
```

```python
np.random.seed(789)

# Setup
regions = ['North', 'South', 'East', 'West']
products = [f'Product_{chr(65+i)}' for i in range(12)]  # Product_A to Product_L
start_date = datetime(2023, 1, 1)

# Generate monthly sales data
data = []
for month_offset in range(24):  # 24 months of data
    month_date = start_date + timedelta(days=30 * month_offset)
    
    for region in regions:
        for product in products:
            units_sold = np.random.randint(50, 800)
            unit_price = round(np.random.uniform(20, 200), 2)
            
            data.append({
                'Date': month_date,
                'Year': month_date.year,
                'Month': month_date.strftime('%B'),
                'Region': region,
                'Product': product,
                'Units_Sold': units_sold,
                'Unit_Price': unit_price,
                'Revenue': round(units_sold * unit_price, 2),
                'Growth_%': round(np.random.uniform(-10, 25), 2)
            })

df = pd.DataFrame(data)
df.to_csv('sales_dashboard_data.csv', index=False)
print(f"✅ Created {len(df):,} monthly sales records across 4 regions and 12 products")
```
